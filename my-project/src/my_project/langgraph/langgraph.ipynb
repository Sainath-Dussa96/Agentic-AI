Hello World using Lang Graph

```python
from typing import TypedDict
from langgraph.graph import StateGraph

class AgentState(TypedDict):
    ping:str

def test_node(state:AgentState):
    """This is a testing One"""
    state['ping']= "Hey " +state["ping"] + ", you are future tech CEO"
    
    return state



graph1=StateGraph(AgentState)

graph1.add_node("testnode1",test_node)

graph1.set_entry_point("testnode1")

graph1.set_finish_point("testnode1")

graph1

app2=graph1.compile()

app2
```

```python
result=app2.invoke({
    "ping":"Sainath"
})

print(result)
```
