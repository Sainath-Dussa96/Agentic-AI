Hello World using Lang Graph

```python
from typing import TypedDict
from langgraph.graph import StateGraph

class AgentState(TypedDict):
    ping:str

def test_node(state:AgentState):
    """This is a testing One"""
    state['ping']= "Hey " +state["ping"] + ", you are future tech CEO"
    
    return state

graph1=StateGraph(AgentState)

graph1.add_node("testnode1",test_node)

graph1.set_entry_point("testnode1")

graph1.set_finish_point("testnode1")

graph1

app2=graph1.compile()

app2
```

 Multi Agent System

```python
from typing import List, TypedDict

class AgentState(TypedDict):
    values:List[int]
    name:str
    result:str
```

```python
def process_values(state:AgentState):
    """This function handles multiple different inputs"""

    state["result"] = f"Hi there {state["name"]}! Your sum={sum(state["values"])}"
    return state
```

```python
graph2=StateGraph(AgentState)

graph2.add_node("node2",process_values)

graph2.set_entry_point("node2")
graph2.set_finish_point("node2")

app3=graph2.compile()
```

```python
from IPython.display import Image, display

display(Image(app3.get_graph().draw_mermaid_png()))
```

```python
answers=app3.invoke(
    {"values":[1,2,3,4],"name":"Sainath"}
)

print(answers["result"])
```

Graph |||

```python
class AgentState(TypedDict):
    finall:str
    name:str
    age:int


def first_node(state:AgentState) -> AgentState:
    """This is the first node of our sequence"""
    state["finall"]=f"Hi {state['name']}!"
    return state


def second_node(state:AgentState) -> AgentState:
    """this is the second node of our sequence"""
    state["finall"]+=f"you are {state["age"]} years old!"
    return state

graph4=StateGraph(AgentState)

graph4.add_node("firstnode",first_node)
graph4.add_node("secondnode",second_node)

graph4.set_entry_point("firstnode")
graph4.add_edge("firstnode","secondnode")
graph4.set_finish_point("secondnode")

app5=graph4.compile()

app5
```

```python
result4=app5.invoke({
    "name":"Sainath", 
    "age":24
})

print(result4)
```
